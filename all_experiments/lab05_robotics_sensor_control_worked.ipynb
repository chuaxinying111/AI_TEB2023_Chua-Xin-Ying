```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import mutual_info_classif
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    accuracy_score,
    confusion_matrix,
    f1_score,
    precision_recall_fscore_support,
)
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, export_text

ROOT = Path.cwd().parent
DATA = ROOT / "all_datasets"
RESULTS = ROOT / "results" / "lab05"
RESULTS.mkdir(parents=True, exist_ok=True)
```

```python
#Ques 1 : Inspeact the data and fix the split

# Name the 24 sensor columns US00 to US23
sensors = [f"US{i:02d}" for i in range(24)]
df = pd.read_csv(DATA / "lab05-sensor_readings_24.csv", header=None, names=sensors + ["action"])

# X = the 24 sensor readings (input), y = the action (what we want to predict) [slope]
X = df[sensors]
y = df.action

print("Shape:", df.shape)
# Class balance: count and share of each action (shows the classes are imbalanced)
display(y.value_counts().rename("count").to_frame().assign(share=lambda t: (t["count"] / len(df)).round(3)))

# Summary stats for the first 6 sensors (transpose so each sensor is a row)
display(X.describe().T[["mean", "std", "min", "max"]].round(2).head(6))

# 70/30 split, stratify=y keeps the class shares equal in train and test, seed 42 makes it repeatable
train_idx, test_idx = train_test_split(df.index, test_size=0.30, random_state=42, stratify=y)

# Label every row as "train" or "test" and save it, so the split is the same every time
split = pd.Series("train", index=df.index, name="split")
split[test_idx] = "test"
split.rename_axis("row_id").to_csv(RESULTS / "split.csv")

# Expect {'train': 3819, 'test': 1637}
print({"train": len(train_idx), "test": len(test_idx)})
```

```python
# Ques 2: Select four sensors using training data only

# Mutual information = how much each sensor tells us about the action
# Only training rows are used, so the test set stays untouched
mi = mutual_info_classif(X.loc[train_idx], y[train_idx], random_state=42)

# Put the scores in a table and sort from most to least informative
feature_table = pd.DataFrame({"sensor": sensors, "mutual_information": mi})
feature_table = feature_table.sort_values("mutual_information", ascending=False)

# Save the full ranking of all 24 sensors
feature_table.to_csv(RESULTS / "feature_scores.csv", index=False)

# Keep the top 4 sensors for the interpretable controller
selected = feature_table.sensor.head(4).tolist()
print("Selected sensors:", selected)
display(feature_table.head(8).round(3))
```

```python
# Ques 3: Fit an interpretable rule controller

# Carve a validation set (20%) out of the TRAINING rows, used only to pick the tree depth
subtrain_idx, val_idx = train_test_split(train_idx, test_size=0.20, random_state=42, stratify=y[train_idx])

# Try depths 2 to 6 and score each by macro F1 on the validation set
depth_results = []
for depth in [2, 3, 4, 5, 6]:
    # min_samples_leaf=20 stops tiny leaves, class_weight="balanced" helps the rare classes
    candidate = DecisionTreeClassifier(max_depth=depth, min_samples_leaf=20, class_weight="balanced", random_state=42)
    # Train on the sub-train rows, using only the 4 selected sensors
    candidate.fit(X.loc[subtrain_idx, selected], y[subtrain_idx])
    # Macro F1 gives every action equal weight, so the rare turns count
    score = f1_score(y[val_idx], candidate.predict(X.loc[val_idx, selected]), average="macro")
    depth_results.append({"max_depth": depth, "validation_macro_f1": score})
depth_table = pd.DataFrame(depth_results)
display(depth_table.round(3))
# Pick the depth with the best validation macro F1 (expect 5)
best_depth = int(depth_table.loc[depth_table.validation_macro_f1.idxmax(), "max_depth"])

# Refit the final tree on ALL training rows using the best depth
rule_model = DecisionTreeClassifier(max_depth=best_depth, min_samples_leaf=20, class_weight="balanced", random_state=42)
rule_model.fit(X.loc[train_idx, selected], y[train_idx])

# Export the tree as readable if-then rules and save them
rules = export_text(rule_model, feature_names=selected)
(RESULTS / "rule_controller.txt").write_text(rules)
print("Selected depth:", best_depth, "| leaves:", rule_model.get_n_leaves())
print(rules)
```

```python
# Ques 4: Fit a reference model

# Random forest on ALL 24 sensors, same training rows
reference = RandomForestClassifier(
    n_estimators=200,
    min_samples_leaf=2,
    class_weight="balanced_subsample",
    random_state=42,
    n_jobs=-1,
).fit(X.loc[train_idx], y[train_idx])
print("Random forest trained on", reference.n_features_in_, "sensors with", reference.n_estimators, "trees")
```

```python
# Ques 5: Evaluate on the held-out data

# True actions for the test rows
y_test = y[test_idx]

# Each model predicts on the test rows (rule tree uses 4 sensors, forest uses all 24)
predictions = {
    "Interpretable rule tree": rule_model.predict(X.loc[test_idx, selected]),
    "Random-forest reference": reference.predict(X.loc[test_idx]),
}

# Collect accuracy and macro precision / recall / F1 for each model
rows = []
for name, prediction in predictions.items():
    precision, recall, f1, _ = precision_recall_fscore_support(y_test, prediction, average="macro")
    rows.append({
        "model": name,
        "accuracy": accuracy_score(y_test, prediction),
        "macro_precision": precision,
        "macro_recall": recall,
        "macro_f1": f1,
    })

# Save and show the comparison table
metrics = pd.DataFrame(rows)
metrics.to_csv(RESULTS / "metrics.csv", index=False)
display(metrics.round(3))

# Class names in a fixed order, so both confusion matrices use the same layout
labels = sorted(y.unique())

# Draw both confusion matrices side by side
fig, axes = plt.subplots(1, 2, figsize=(11, 4.2))
for ax, (name, prediction) in zip(axes, predictions.items()):
    cm = confusion_matrix(y_test, prediction, labels=labels)
    ConfusionMatrixDisplay(cm, display_labels=labels).plot(ax=ax, colorbar=False, xticks_rotation=30)
    ax.set_title(name)
plt.tight_layout()
plt.savefig(RESULTS / "confusion_matrices.png", dpi=160)

# Per-class precision, recall and F1 for the rule tree only
precision, recall, f1, _ = precision_recall_fscore_support(
    y_test, predictions["Interpretable rule tree"], labels=labels
)
per_class = pd.DataFrame({"precision": precision, "recall": recall, "f1": f1}, index=labels)
per_class.to_csv(RESULTS / "rule_per_class.csv")
display(per_class.round(3))

# The action with the lowest recall is the hardest (expect Slight-Right-Turn)
print("Hardest action for the rule controller (lowest recall):", per_class.recall.idxmin())
```

```python
# Ques 6: Test robustness to sensor noise

# Random generator with a fixed seed, so the noise is the same every run
rng = np.random.default_rng(42)

# Standard deviation of each selected sensor on the TRAINING rows (noise size is based on this)
train_sd = X.loc[train_idx, selected].std().to_numpy()

# Clean test readings of the 4 selected sensors as a number array
clean = X.loc[test_idx, selected].to_numpy(float)

# Add zero-mean Gaussian noise at 0%, 2%, 5% and 10% of the training SD, then re-score the rule tree
noise_rows = []
for fraction in [0.0, 0.02, 0.05, 0.10]:
    noisy = clean + rng.normal(0.0, fraction * train_sd, size=clean.shape)
    # Put the column names back so the tree accepts the data
    prediction = rule_model.predict(pd.DataFrame(noisy, columns=selected))
    noise_rows.append({
        "noise_fraction_of_train_sd": fraction,
        "macro_f1": f1_score(y_test, prediction, average="macro"),
    })
noise = pd.DataFrame(noise_rows)
noise.to_csv(RESULTS / "noise_robustness.csv", index=False)
display(noise.round(3))

# Line plot: macro F1 against noise level
ax = noise.plot(x="noise_fraction_of_train_sd", y="macro_f1", marker="o", legend=False, figsize=(6, 4))
ax.set(
    xlabel="Gaussian noise (fraction of training SD)",
    ylabel="Macro F1",
    title="Rule-controller robustness",
)
plt.tight_layout()
plt.savefig(RESULTS / "noise_robustness.png", dpi=160)
print("Saved:", sorted(p.name for p in RESULTS.iterdir()))
```
