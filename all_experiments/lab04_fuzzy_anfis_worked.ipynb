```python
from itertools import product
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from scipy.optimize import minimize
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
ROOT = Path.cwd().parent
DATA = ROOT / "all_datasets"
RESULTS = ROOT / "results" / "lab04"
RESULTS.mkdir(parents=True, exist_ok=True)
```

```python
# Ques 1: Clean the data
df = pd.read_csv(DATA / "lab04-AirQuality.csv", sep=";", decimal=",")
df = df.dropna(axis="columns", how="all").dropna(how="all") # empty unnamed columns, empty tail rows
df = df.replace(-200, np.nan)
print("Rows:", len(df))
display(df[["T", "RH", "CO(GT)", "PT08.S1(CO)", "PT08.S2(NMHC)"]].notna().sum().rename("usable rows").
to_frame())

#alpha=transparency, so crowded areas look darker
ax = df.plot.scatter(x="T", y="RH", s=7, alpha=0.25, figsize=(6, 4))
ax.set(xlabel="Temperature (deg C)", ylabel="Relative humidity (%)", title="Operating conditions")
plt.tight_layout()
plt.savefig(RESULTS / "temperature_humidity.png", dpi=160)
```

```python
# Ques 2 : Build the fuzyy controller
def trimf(x, a, b, c):
    """Triangle with feet a and c and peak b. a == b or b == c makes a shoulder."""
    x = np.asarray(x, dtype=float)
    # left slope:rises from 0(at a) to 1(at b)
    #if a==b there is no slope, so it's just 1 everywhere(shoulder)
    #np.clip keeps values between 0 and 1
    left = np.ones_like(x) if a == b else np.clip((x - a) / (b - a), 0, 1)

    #righ slope:falls from 1(at b) to 0 (at c)
    #same shoulder trick if b==c
    right = np.ones_like(x) if b == c else np.clip((c - x) / (c - b), 0, 1)

    #the triangle is whichever side is lower (min of the two slopes)
    return np.minimum(left, right)

# the "universe" of each variable = all values it can take.
# T: 0-45 C, RH: 10-90 %, ventilation: 0-100 %
# many points so the curves and the centroid are smooth
T_range, RH_range, V_range = np.linspace(0, 45, 451), np.linspace(10, 90, 401), np.linspace(0, 100, 501)

#temperature sets: cold and hot are shoulders, mild is real triangle
T_sets = {"cold": trimf(T_range, 0, 0, 18), "mild": trimf(T_range, 12, 22, 32), "hot": trimf(T_range, 26, 45, 45)}

#humidity sets
RH_sets = {"dry": trimf(RH_range, 10, 10, 40), "comfortable": trimf(RH_range, 30, 50, 70), "humid": trimf(RH_range, 60, 90, 90)}

#output sets: how much ventilation we want
V_sets = {"low": trimf(V_range, 0, 0, 45), "medium": trimf(V_range, 25, 50, 75), "high": trimf(V_range, 55,100, 100)}
def degree(universe, membership, value):
    """Membership degree of one crisp value."""
    return float(np.interp(value, universe, membership))

def ventilation(temperature, humidity):
    # Stage 1 - fuzzification: turn each crisp input into a degree
    # for every set, e.g. t = {"cold": 0.17, "mild": 0.30, "hot": 0}
    t = {name: degree(T_range, mf, temperature) for name, mf in T_sets.items()}
    h = {name: degree(RH_range, mf, humidity) for name, mf in RH_sets.items()}
    # Stage 2 - rule base : AND = min (weakest condition limits it),
    # rules with the same conclusion are joined with max (best one wins)

    ## low if: cold AND dry, OR cold AND comfortable
    low = max(min(t["cold"], h["dry"]), min(t["cold"], h["comfortable"]))
    
    # medium if: mild AND dry, OR mild AND comfortable, OR cold AND humid
    medium = max(min(t["mild"], h["dry"]), min(t["mild"], h["comfortable"]), min(t["cold"], h["humid"]))
    
    # high if: hot, OR humid (either one is enough)
    high = max(t["hot"], h["humid"])
   
    # Stage 3 - inference: clip each output set at its rule strength
    # (a rule firing at 0.3 can only "fill" the set up to 0.3),
    # then join all the clipped shapes with max
    output = np.maximum.reduce([np.minimum(low, V_sets["low"]), np.minimum(medium, V_sets["medium"]), np.
    minimum(high, V_sets["high"])])

    # Stage 4 - defuzzification: centroid = balance point of the shape
    # = sum(x * height) / sum(height). gives one crisp number in %
    return float(np.sum(V_range * output) / np.sum(output))

fig, axes = plt.subplots(1, 3, figsize=(12, 3.4))

# loop through panels and draw every set as a line
for ax, (title, universe, sets) in zip(axes, [("Temperature", T_range, T_sets), ("Humidity", RH_range,RH_sets), ("Ventilation", V_range, V_sets)]):
    for name, membership in sets.items():
        ax.plot(universe, membership, label=name)
    ax.set_title(title)
    ax.legend()
plt.tight_layout()
plt.savefig(RESULTS / "fuzzy_memberships.png", dpi=160)
```

```python
#Ques 3: Validate the rule base
# T_C = temperature, RH_pct = humidity, expected = the band we expect
fixed = pd.DataFrame({"T_C": [10, 20, 35, 25, 40], "RH_pct": [30, 50, 50, 80, 80],"expected": ["low", "medium", "high", "high", "high"]})
fixed["ventilation_pct"] = [ventilation(t, h) for t, h in zip(fixed.T_C, fixed.RH_pct)]
fixed.to_csv(RESULTS / "fuzzy_fixed_points.csv", index=False)
display(fixed.round(1))

temperatures, humidities = np.linspace(0, 45, 46), np.linspace(10, 90, 41)
surface = np.array([[ventilation(t, h) for t in temperatures] for h in humidities])

fig, ax = plt.subplots(figsize=(7, 4.5))
contour = ax.contourf(temperatures, humidities, surface, levels=15)
fig.colorbar(contour, label="Ventilation demand (%)")
ax.set(xlabel="Temperature (deg C)", ylabel="Relative humidity (%)", title="Fuzzy control surface")
plt.tight_layout()
plt.savefig(RESULTS / "fuzzy_control_surface.png", dpi=160)
```

```python
#Ques 4: Prepare the ANFIS task
task = df[["CO(GT)", "PT08.S1(CO)", "PT08.S2(NMHC)"]].dropna()

# X = inputs (the two metal-oxide sensors), as a plain numpy array
X = task[["PT08.S1(CO)", "PT08.S2(NMHC)"]].to_numpy()

# y = what we want to predict (reference CO concentration)
y = task["CO(GT)"].to_numpy()

# first split: 70% train, 30% "rest". We split row NUMBERS (0, 1, 2)
# not the data itself, so the same indices can be reused for X and y.
# random_state=42 makes the shuffle repeatable
train, rest = train_test_split(np.arange(len(task)), test_size=0.30, random_state=42)

# second split: cut the 30% in half -> 15% validation, 15% test
val, test = train_test_split(rest, test_size=0.50, random_state=42)

# make a label for every row, "train" by default
split = pd.Series("train", index=task.index, name="split")
split.iloc[val], split.iloc[test] = "validation", "test"

# save which row went where, so the split can be reproduced and checked
split.rename_axis("source_row").to_csv(RESULTS / "split.csv")

# learn the mean and std from TRAINING rows only
scaler = StandardScaler().fit(X[train])

# apply that same scaling to ALL rows (train, validation, test)
Xs = scaler.transform(X)
print({"usable_rows": len(task), "train": len(train), "validation": len(val), "test": len(test)})
```

```python

# Ques 5: Fit two model classess
class SugenoANFIS:
    """First-order Sugeno ANFIS: Gaussian memberships, linear rule outputs."""

    def __init__(self, n_mfs=2, ridge=1e-4, maxiter=30):
        self.n_mfs = n_mfs
        self.ridge = ridge
        self.maxiter = maxiter

    def _weights(self, X, theta):
        """Layers 1-3: memberships -> rule strengths -> normalised strengths."""
        # number of inputs (2 sensors)
        n_features = X.shape[1]
        k = n_features * self.n_mfs
        centers = theta[:k].reshape(n_features, self.n_mfs)
        widths = np.exp(theta[k:]).reshape(n_features, self.n_mfs)

        # LAYER 1: Gaussian membership of every sample in every set
        membership = []
        for j in range(n_features):
            z = (X[:, [j]] - centers[j]) / widths[j]
            membership.append(np.exp(-0.5 * z ** 2))

        # LAYER 2: one rule per combination of sets
        columns = []
        for rule in product(range(self.n_mfs), repeat=n_features):
            # pick the membership of each input for this rule
            parts = [membership[j][:, m] for j, m in enumerate(rule)]
            # AND = multiply them -> firing strength of this rule
            columns.append(np.prod(parts, axis=0))
        strength = np.column_stack(columns)

        # LAYER 3: normalise so each row of strengths sums to 1
        return strength / (strength.sum(axis=1, keepdims=True) + 1e-12)

    def _design(self, X, theta):
        """Layer 4: every rule contributes weight x (1, x1, x2)."""
        weights = self._weights(X, theta)
        inputs = np.column_stack([np.ones(len(X)), X])
        blocks = [weights[:, [r]] * inputs for r in range(weights.shape[1])]
        return np.hstack(blocks)

    def _solve_outputs(self, X, y, theta):
        Z = self._design(X, theta)
        A = Z.T @ Z + self.ridge * np.eye(Z.shape[1])
        return np.linalg.solve(A, Z.T @ y)

    def fit(self, X, y):
        n_features = X.shape[1]
        # starting centres: spread between the 25th and 75th percentile
        # of each input, so they begin where the data actually is
        centers = np.array([
            np.quantile(X[:, j], np.linspace(0.25, 0.75, self.n_mfs))
            for j in range(n_features)
        ])
        widths = np.full((n_features, self.n_mfs), max(X.std(), 0.5))
        theta0 = np.r_[centers.ravel(), np.log(widths).ravel()]

        def training_mse(theta):
            Z = self._design(X, theta)
            p = self._solve_outputs(X, y, theta)
            return np.mean((y - Z @ p) ** 2)

        bounds = [(-3, 3)] * centers.size
        bounds += [(np.log(0.15), np.log(3))] * widths.size

        result = minimize(
            training_mse, theta0, method="L-BFGS-B",
            bounds=bounds, options={"maxiter": self.maxiter},
        )
        self.theta_ = result.x
        self.outputs_ = self._solve_outputs(X, y, self.theta_)
        return self

    def predict(self, X):
        # LAYER 5: sum of all rule outputs = design matrix @ coefficients
        return self._design(X, self.theta_) @ self.outputs_


def scores(name, model):
    row = {"model": name}
    for part, rows in [("validation", val), ("test", test)]:
        prediction = model.predict(Xs[rows])
        row[f"{part}_rmse"] = mean_squared_error(y[rows], prediction) ** 0.5
        row[f"{part}_mae"] = mean_absolute_error(y[rows], prediction)
    return row


linear = LinearRegression().fit(Xs[train], y[train])
candidates = {n: SugenoANFIS(n_mfs=n).fit(Xs[train], y[train]) for n in [2, 3]}

rows = [scores("Linear regression", linear)]
rows += [scores(f"ANFIS-{n}MF", m) for n, m in candidates.items()]
metrics = pd.DataFrame(rows)
metrics.to_csv(RESULTS / "metrics.csv", index=False)
display(metrics.round(4))

by_model = metrics.set_index("model")
best_n = min(candidates, key=lambda n: by_model.loc[f"ANFIS-{n}MF", "validation_rmse"])
anfis = candidates[best_n]
print(f"Selected by validation RMSE: ANFIS-{best_n}MF")
```

```python
# Ques 6: Evaluate once on the test data
fig, axes = plt.subplots(1, 2, figsize=(9, 4), sharex=True, sharey=True)

models = [
    ("Linear baseline", linear),
    (f"ANFIS-{best_n}MF", anfis),
]

top = y[test].max()

# loop: one panel per model
for ax, (name, model) in zip(axes, models):
    # predict on the TEST rows (the first time we use them for scoring)
    prediction = model.predict(Xs[test])
    ax.scatter(y[test], prediction, s=10, alpha=0.35)
    ax.plot([0, top], [0, top], "--", linewidth=1)
    ax.set(title=name, xlabel="Actual CO(GT)", ylabel="Predicted CO(GT)")

plt.tight_layout()
plt.savefig(RESULTS / "predicted_vs_actual.png", dpi=160)
plt.show()

anfis_rmse = by_model.loc[f"ANFIS-{best_n}MF", "test_rmse"]
linear_rmse = by_model.loc["Linear regression", "test_rmse"]
# relative improvement: how much smaller ANFIS's error is vs linear.
improvement = 1 - anfis_rmse / linear_rmse

print(f"Linear test RMSE: {linear_rmse:.4f}")
print(f"ANFIS-{best_n}MF test RMSE: {anfis_rmse:.4f}")
print(f"Test RMSE improvement over linear regression: {improvement:.1%}")
print("Saved:", sorted(p.name for p in RESULTS.iterdir()))
```
